# Numpy

```python
import numpy as np

A = np.array([1, 2, 3])
```

```python
print(A)
B = np.arange(2).reshape(2, 1)
print(B)
print(A + B)
```

```python
# hstack

A = np.array([1, 2, 3]).reshape(3, 1)
B = np.array([4, 5, 6, 7, 8, 9]).reshape(3, 2)
print(np.hstack((A, B)))
print(np.hstack((A, B)).shape)
```

```python
# vstack
A = np.array([1, 2, 3]).reshape(1, 3)
B = np.array([4, 5, 6, 7, 8, 9]).reshape(2, 3)
print(np.vstack((A, B)))
print(np.vstack((A, B)).shape)
```

```python
# sort

A = np.arange(6).reshape(2, 3)
print(A)
print(np.sort(A)[::-1])
```

```python
# Universal functions are very fast
A = np.random.rand(1000000)
%timeit sum(A)
%timeit np.sum(A)
```
