```python
# Slicing

A = np.arange(6)
print(A)
B = A[2:4]
print(B)
B[1] = 100
print(B)
print(A)
C = A[2:4].copy()
print(C)
```

```python
A = np.arange(10, 20)
idx = np.argwhere(A == 10)  # to find the index of 10 in the array
print(idx[0][0])

A = A.reshape(2, 5)
print(A)
idex2 = np.argwhere(A == 18)
print(idex2)
```

```python
X = np.round(np.random.rand(100) * 10)  #generated a random array of float with out decimal

Y = X.astype(int)
print(Y)
print(Y.reshape(2, 50).T)  # T for transpose
```

```python
ABC = np.arange(100, 0, -1)
print(ABC)
ABC.sort()
print(ABC)
ABC = ABC.reshape(25, 4)
ABC.sort(axis=0)
print(ABC)
```

```python
#Masking Index Array

A = np.array([12, 45, 67, 75, 1, 25])
print(A)
print(A[[1, 3, 5]])  # Pick the values for the mentioned indices only
print(A[[True, False, False, True, True,
         False]])  # Pick the values for the indices whereever true is mentioned

print(A[A > 45])
```

```python
A = np.array([[12, -7, 56, 1], [13, -8, 57, 2]])
print(A[0][1:3])
print(A[A > 13])
```

```python
A = np.arange(100)
print(A[(A > 25) & (A <= 90)])

print(type(A))

B = [A + 20]  # this will create a list. This is called broadcasting
print(type(B))
print(B)
```

```python
# Broadcasting another example

A = np.arange(100).reshape(25, 4)
print(A + [10, 11, 12, 13])
```
