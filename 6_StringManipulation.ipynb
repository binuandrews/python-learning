# String Manipulation

```python
# String manipulation
s1 = "The price is"
s2 = "$"

print(s1+" "+str(200)+s2)

multilineString = """ This string spanned across
                Multiple lines  """
print(multilineString)
```

```python
# [start index: end index : step]

s = "How are you and who are you?"

print(s)
print(s[5])
print(s[5:len(s)])  # Last index is not included
print(s[5:12])


print("******   "+s[-8:-3])
print("******   "+s[-3:-8:-1])
print("******   "+s[::-1]) # To reverse
```

```python
s = "          How are you ????and who are you?       "
print(s.strip())
print(s)
print(s.lower())
print(s.upper())
print(s.replace("you", "we"))
print(s.replace("????", "    "))
```

```python
str = "a123, 456, 878, abc, PQR, xyz"
print(str.capitalize()) # capitalize if first chat is an alphabet
L = str.split(", ")
print(L)

for item in L:
  print(item)
```

```python
if "abc" in "pqr abc xyz":
  print("String contains abc")
else:
  print("String does not contain abc")

print("abcd" in "pqr abc xyz")
print("abc" in "pqr abc xyz")
print("abc" not in "pqr abc xyz")

```

```python
str1 = "ABC"
str2 = "abc"
print(str1 == str2)
print(str1 > str2)
print(str1 < str2)
print(str1 < str2)
print(str1 == str2.upper())
```

```python
# Escape Sequences

print("He is a \"good\" boy")
print('He is a \"good\" boy')
print('He is a "good" boy')
print('He is a \"good\" \nboy')

print(r"c:\newdir\abc")   # r represents raw here. N escape sequences will be considered
```
