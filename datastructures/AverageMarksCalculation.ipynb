```python

def getDataFromUser():
  myDict = {}
  while True:
    studentId = input("Enter student ID: ")
    studetMarks = input("Enter student Marks separated by comma: ")
    myDict.update({studentId: studetMarks.split(',')})
    if input("Enter no to quit: ").lower() == "no":
      break


  return myDict


myDict = getDataFromUser()
print(myDict)

```

```python


def printAverageMarks(myDict):
  for key, marks in myDict.items():
    sum = 0
    for mark in marks:
      sum += int(mark)

    print("Student ID", key, "Averge mark is ", sum/len(marks))

printAverageMarks(myDict)
```
