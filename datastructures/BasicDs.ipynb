```python
myList = [23, 12, 2.0, False, 3+4j, "Binu", True, 12, (12, "Test")]
myTuple = (23, 12, 2.0, False, 3+4j, "Binu", True, 12, (12, "Test"))
mySet = {23, 12, 2.0, False, 3+4j, "Binu", True, 12, (12, "Test")}
myDictionary = {23: "Twenty Three", "Twelve": 12, 13: "Thirteen", 13: "Fourteen"}

print(type(myList),"___________",myList)
print(type(myTuple),"___________",myTuple)
print(type(mySet),"___________",mySet)
print(type(myDictionary),"___________",myDictionary)

print("****************************************")
print(myList[0], myList[len(myList)-1])
print(myTuple[0], myTuple[len(myTuple)-1])
print("****************************************")
for item in mySet:
  print(item)
print("****************************************")
print("Binu" in mySet)
print((12, "Test") in mySet)
print(myDictionary["Twelve"], myDictionary[23])
```
