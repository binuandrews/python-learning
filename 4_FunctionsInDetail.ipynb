# Functions

```python
#Function basic demo

def subtract(a,b):
    """  This function adds two numbers   """
    print("Calculating difference between of {} and {}".format(a,b))
    return a-b


num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))

print("Result is ",subtract(b = num2, a = num1))    # Order specified explicitly
print("Result is ",subtract(num1, num2))    # Default order maintained
print("Result is ",subtract(num2, num1))    # Order reversed

```

```python
# return statement can return multiple values.

def divideEachBy2(a, b):
  return a/2, b/2

print(divideEachBy2(10, 5))
print(type(divideEachBy2(10, 5)))
```

```python
# Variable number of args to a function
def printAllNumbers(**abc):    # Arg is like dictionary, if we pass **abc
  print("Type of input variable  ",type(abc))
  for item in abc.items():
    print(item[0], item[1])

  print("***********")
  for key, value in abc.items():
    print(key, value, type(value))

  print("***********")
  for key in abc:
    print(key, abc[key], type(abc[key]))


printAllNumbers(abc=20, b=30, d=100, name="Binu Andrews")
```
